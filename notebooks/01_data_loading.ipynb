## Data Loading

````python
def load_sales_data(filepath=None):
    """Generate synthetic sales data for testing.
    
    Args:
        filepath: Ignored. Synthetic data is generated.
        
    Returns:
        DataFrame with sales data
    """
    import pandas as pd
    import numpy as np

    np.random.seed(42)
    n_orders = 1000
    categories = ['Electronics', 'Clothing', 'Home', 'Books', 'Sports']
    products = [f"P{str(i).zfill(3)}" for i in range(1, 101)]
    product_names = [f"Product_{i}" for i in range(1, 101)]
    customers = [f"CUST{str(i).zfill(4)}" for i in range(1, 201)]
    regions = ['North', 'South', 'East', 'West']

    data = {
        'order_id': np.arange(1, n_orders + 1),
        'product_id': np.random.choice(products, n_orders),
        'product_name': np.random.choice(product_names, n_orders),
        'category': np.random.choice(categories, n_orders),
        'quantity': np.random.randint(1, 6, n_orders),
        'price': np.round(np.random.uniform(5.0, 500.0, n_orders), 2),
        'order_date': pd.date_range('2024-01-01', periods=n_orders, freq='D'),
        'customer_id': np.random.choice(customers, n_orders),
        'region': np.random.choice(regions, n_orders)
    }
    pdf = pd.DataFrame(data)
    df = spark.createDataFrame(pdf)
    return df
````

````python
def calculate_summary_metrics(df):
    """Calculate summary metrics from sales data.
    
    TODO: This function needs to be enhanced!
    
    User A: Add revenue metrics
    - total_revenue
    - average_order_value
    - revenue_by_category
    
    User B: Add customer metrics
    - unique_customers
    - average_quantity_per_order
    - top_customers (top 5 by spend)
    
    Best Practice: Document parameters and return values!
    
    Args:
        df: Sales DataFrame
        
    Returns:
        dict: Summary metrics
    """
    total_orders = df.count()
    df_with_revenue = df.withColumn("order_revenue", df["quantity"] * df["price"])
    total_revenue = df_with_revenue.agg({"order_revenue": "sum"}).collect()[0][0]
    average_order_value = total_revenue / total_orders if total_orders > 0 else 0
    revenue_by_category = (
        df_with_revenue.groupBy("category")
        .agg({"order_revenue": "sum"})
        .toPandas()
        .set_index("category")["sum(order_revenue)"]
        .to_dict()
    )

    summary = {
        'total_orders': total_orders,
        'total_revenue': total_revenue,
        'average_order_value': average_order_value,
        'revenue_by_category': revenue_by_category,
        # More metrics will be added by User B
    }
    return summary
````

````python
import matplotlib.pyplot as plt

# Calculate summary metrics
metrics = calculate_summary_metrics(df)

# Print summary metrics
print("\nSummary Metrics:")
for key, value in metrics.items():
    print(f"  {key}: {value}")

# Plot revenue by category
revenue_by_category = metrics.get("revenue_by_category", {})
if revenue_by_category:
    plt.figure(figsize=(8, 5))
    plt.bar(revenue_by_category.keys(), revenue_by_category.values(), color='skyblue')
    plt.xlabel("Category")
    plt.ylabel("Revenue")
    plt.title("Revenue by Category")
    plt.tight_layout()
    plt.show()
````
